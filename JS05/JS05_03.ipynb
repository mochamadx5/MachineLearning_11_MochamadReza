```python
# 1. Muat ulang data asli
df = pd.read_csv('spam.csv', encoding='latin-1')

# 2. Hapus 3 kolom terakhir yang tidak digunakan
df = df.drop(df.iloc[:, 2:], axis=1)

# 3. Rename kolom agar konsisten
df = df.rename(columns={'v1': 'Labels', 'v2': 'SMS'})

# 4. Map label ke nilai numerik (spam: 1, ham: 0)
new_labels = {'spam': 1, 'ham': 0}
df['Labels'] = df['Labels'].map(new_labels)

# 5. Tampilkan hasil untuk memastikan tidak ada NaN
display(df.head(5))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# Split data baru yang sudah bersih
X_train, X_test, y_train, y_test = train_test_split(df['SMS'], df['Labels'], test_size=0.2, random_state=50)

# Inisiasi dan transformasi teks menggunakan CountVectorizer
bow = CountVectorizer()
X_train = bow.fit_transform(X_train)
X_test = bow.transform(X_test)

print(f'Jumlah fitur kata: {len(bow.get_feature_names_out())}')
print(f'Dimensi data latih: {X_train.shape}')
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# inisiasi MultinomialNB
mnb = MultinomialNB()

# fit model
mnb.fit(X_train, y_train)

# prediksi dengan data training
y_pred_train = mnb.predict(X_train)

# evaluasi akurasi data training
acc_train = accuracy_score(y_train, y_pred_train)

#prediksi dengan data training
y_pred_test = mnb.predict(X_test)

# evaluasi akurasi data training
acc_test = accuracy_score(y_test, y_pred_test)

# print hasil evaluasi
print(f'Hasil akurasi data train: {acc_train}')
print(f'Hasil akurasi data test: {acc_test}')
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Muat ulang dataset
df = pd.read_csv('spam.csv', encoding='latin-1')

# Hapus kolom tidak penting dan rapikan nama kolom
df = df.drop(df.iloc[:, 2:], axis=1)
df = df.rename(columns={'v1': 'Labels', 'v2': 'SMS'})

# Bersihkan whitespace jika ada pada teks label
df['Labels'] = df['Labels'].astype(str).str.strip()

# Encoding label secara aman
new_labels = {'spam': 1, 'ham': 0}
df['Labels'] = df['Labels'].map(new_labels)

# Pastikan tidak ada NaN yang terbentuk
print("Jumlah NaN setelah encoding:", df['Labels'].isnull().sum())
display(df.head())
```

```python
# Split data menjadi train dan test set
X_train, X_test, y_train, y_test = train_test_split(df['SMS'], df['Labels'], test_size=0.2, random_state=50)

# Ekstraksi Fitur dengan CountVectorizer
bow = CountVectorizer()
X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)

# Training Model MultinomialNB
mnb = MultinomialNB()
mnb.fit(X_train_bow, y_train)

# Evaluasi Hasil
y_pred_train = mnb.predict(X_train_bow)
acc_train = accuracy_score(y_train, y_pred_train)

y_pred_test = mnb.predict(X_test_bow)
acc_test = accuracy_score(y_test, y_pred_test)

print(f'Hasil akurasi data train: {acc_train:}')
print(f'Hasil akurasi data test: {acc_test:}')
```
